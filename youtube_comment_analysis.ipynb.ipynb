```python
import sys
import subprocess
import warnings
warnings.filterwarnings("ignore")

# Install missing packages quietly
packages = ["pandas", "numpy", "scikit-learn", "nltk", "emoji", "matplotlib"]

for package in packages:
    try:
        if package == "scikit-learn":
            import sklearn
        else:
            __import__(package)
    except ImportError:
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", "-q", package]
        )

import pandas as pd
import numpy as np
import re
import emoji
import nltk
import matplotlib.pyplot as plt

from collections import Counter
from tkinter import Tk, filedialog

from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer


# -------------------- 2. NLTK RESOURCES -------------------------

try:
    nltk.data.find("corpora/stopwords")
except LookupError:
    nltk.download("stopwords", quiet=True)


# -------------------- 3. SELECT CSV FILE -------------------------

print("=" * 70)
print(r"C:\Users\rupes\Downloads\full_dataset_labeled.csv")
print("=" * 70)

root = Tk()
root.withdraw()
root.attributes("-topmost", True)

file_path = filedialog.askopenfilename(
    title="Select your YouTube Comments CSV file",
    filetypes=[
        ("CSV files", "*.csv"),
        ("All files", "*.*")
    ]
)

root.destroy()

if not file_path:
    raise FileNotFoundError(
        "No CSV file was selected. Please run this cell again and select your dataset."
    )

print("\nSelected file:")
print(file_path)


# -------------------- 4. LOAD DATASET ----------------------------

try:
    df = pd.read_csv(
        file_path,
        encoding="utf-8-sig",
        low_memory=False
    )
except UnicodeDecodeError:
    df = pd.read_csv(
        file_path,
        encoding="latin1",
        low_memory=False
    )

print("\nDataset loaded successfully!")


# -------------------- 5. BASIC INFORMATION ----------------------

print("\n" + "=" * 70)
print("INITIAL DATASET INFORMATION")
print("=" * 70)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nColumn names:")
for i, col in enumerate(df.columns):
    print(f"{i+1}. {col}")


# -------------------- 6. DISPLAY FIRST RECORDS -------------------

print("\nFirst 5 records:")
display(df.head())


# -------------------- 7. AUTOMATIC TEXT COLUMN DETECTION ---------

possible_text_columns = [
    "text",
    "comment",
    "comments",
    "comment_text",
    "commentText",
    "content",
    "review",
    "sentence",
    "message"
]

text_column = None

# Exact match first
for col in possible_text_columns:
    if col in df.columns:
        text_column = col
        break

# Partial match if exact match not found
if text_column is None:
    for col in df.columns:
        col_lower = str(col).lower()

        if any(
            keyword in col_lower
            for keyword in ["comment", "text", "content", "review", "sentence"]
        ):
            text_column = col
            break

if text_column is None:
    raise ValueError(
        "\nCould not automatically identify the comment column.\n"
        "Please check the column names printed above."
    )

print("\nDetected comment/text column:", text_column)


# -------------------- 8. SENTIMENT COLUMN DETECTION -------------

possible_sentiment_columns = [
    "sentiment",
    "sentiment_label",
    "sentiment_category",
    "label",
    "polarity",
    "emotion"
]

sentiment_column = None

for col in possible_sentiment_columns:
    if col in df.columns:
        sentiment_column = col
        break

if sentiment_column is None:
    for col in df.columns:
        col_lower = str(col).lower()

        if any(
            keyword in col_lower
            for keyword in ["sentiment", "polarity", "label", "emotion"]
        ):
            sentiment_column = col
            break

if sentiment_column:
    print("Detected sentiment column:", sentiment_column)
else:
    print("No sentiment column detected.")


# -------------------- 9. LANGUAGE COLUMN DETECTION --------------

possible_language_columns = [
    "language",
    "lang",
    "language_label",
    "lang_label",
    "language_category",
    "stage3_category",
    "category"
]

language_column = None

for col in possible_language_columns:
    if col in df.columns:
        language_column = col
        break

if language_column is None:
    for col in df.columns:
        col_lower = str(col).lower()

        if "language" in col_lower or col_lower == "lang":
            language_column = col
            break

if language_column:
    print("Detected language column:", language_column)
else:
    print("No dedicated language column detected.")


# -------------------- 10. CREATE WORKING DATASET ----------------

data = df.copy()

data["original_text"] = (
    data[text_column]
    .fillna("")
    .astype(str)
    .str.strip()
)

# Sentiment standardization
if sentiment_column:
    data["sentiment"] = (
        data[sentiment_column]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.strip()
    )


# -------------------- 11. REMOVE EMPTY COMMENTS -----------------

before = len(data)

data = data[
    data["original_text"].str.strip() != ""
].copy()

print("\nEmpty comments removed:", before - len(data))


# -------------------- 12. REMOVE DUPLICATES ---------------------

before = len(data)

data = data.drop_duplicates(
    subset=["original_text"]
).copy()

print("Duplicate comments removed:", before - len(data))


# -------------------- 13. LANGUAGE DISTRIBUTION -----------------

if language_column:

    print("\n" + "=" * 70)
    print("LANGUAGE DISTRIBUTION BEFORE FILTERING")
    print("=" * 70)

    print(
        data[language_column]
        .fillna("Unknown")
        .astype(str)
        .value_counts()
        .to_string()
    )

    print(
        "\nNOTE: No language category has been automatically deleted."
    )

    print(
        "This is intentional so Gujarati/Hinglish categories are not "
        "mistakenly removed."
    )


# -------------------- 14. SENTIMENT DISTRIBUTION ----------------

if sentiment_column:

    print("\n" + "=" * 70)
    print("SENTIMENT DISTRIBUTION")
    print("=" * 70)

    print(
        data["sentiment"]
        .value_counts()
        .to_string()
    )


# -------------------- 15. TEXT CLEANING FUNCTIONS ---------------

def remove_urls(text):
    text = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text
    )
    return text


def remove_mentions(text):
    text = re.sub(
        r"@\w+",
        " ",
        text
    )
    return text


def process_hashtags(text):
    # Preserve hashtag word instead of deleting it
    text = re.sub(
        r"#(\w+)",
        r"\1",
        text
    )
    return text


def convert_emojis(text):
    # Convert emojis into meaningful words
    return emoji.demojize(
        text,
        delimiters=(" ", " ")
    )


def normalize_repeated_characters(text):
    # Example:
    # soooo gooddddd -> soo good
    return re.sub(
        r"(.)\1{2,}",
        r"\1\1",
        text
    )


def remove_special_characters(text):
    # Preserve English/Romanized Hindi characters and numbers
    text = re.sub(
        r"[^a-zA-Z0-9\s]",
        " ",
        text
    )
    return text


def normalize_spaces(text):
    text = re.sub(
        r"\s+",
        " ",
        text
    )
    return text.strip()


# -------------------- 16. COMPLETE PREPROCESSING ----------------

def preprocess_text(text):

    text = str(text)

    # Emoji -> words
    text = convert_emojis(text)

    # URLs
    text = remove_urls(text)

    # Mentions
    text = remove_mentions(text)

    # Hashtags
    text = process_hashtags(text)

    # Lowercase
    text = text.lower()

    # Repeated character normalization
    text = normalize_repeated_characters(text)

    # Convert emoji formatting
    text = text.replace(":", " ")
    text = text.replace("_", " ")

    # Special character removal
    text = remove_special_characters(text)

    # Whitespace normalization
    text = normalize_spaces(text)

    return text


print("\nApplying text preprocessing...")


data["cleaned_text"] = data[
    "original_text"
].apply(preprocess_text)


# -------------------- 17. REMOVE EMPTY CLEANED TEXT -------------

data = data[
    data["cleaned_text"].str.strip() != ""
].copy()


# -------------------- 18. TOKENIZATION ---------------------------

data["tokens"] = data[
    "cleaned_text"
].apply(
    lambda x: x.split()
)


# -------------------- 19. STOP WORD PROCESSING -------------------

english_stopwords = set(
    stopwords.words("english")
)

# Keep sentiment-important negation words
important_negations = {
    "no",
    "not",
    "never",
    "nor",
    "neither"
}

english_stopwords = (
    english_stopwords
    - important_negations
)


def remove_stopwords(tokens):

    return [
        word
        for word in tokens
        if word not in english_stopwords
    ]


data["tokens_clean"] = data[
    "tokens"
].apply(remove_stopwords)


# -------------------- 20. RECONSTRUCT PROCESSED TEXT -----------

data["processed_text"] = data[
    "tokens_clean"
].apply(
    lambda x: " ".join(x)
)


# -------------------- 21. WORD COUNT -----------------------------

data["word_count"] = data[
    "processed_text"
].apply(
    lambda x: len(x.split())
)


data["character_count"] = data[
    "processed_text"
].apply(len)


# Remove completely empty records
data = data[
    data["processed_text"].str.strip() != ""
].copy()


# -------------------- 22. SHOW PREPROCESSING EXAMPLES -----------

print("\n" + "=" * 70)
print("PREPROCESSING EXAMPLES")
print("=" * 70)

example_columns = [
    "original_text",
    "cleaned_text",
    "processed_text"
]

display(
    data[example_columns].head(15)
)


# -------------------- 23. DATASET STATISTICS ---------------------

print("\n" + "=" * 70)
print("FINAL DATASET STATISTICS")
print("=" * 70)

print("Total comments:", len(data))
print("Total columns :", len(data.columns))

print("\nComment length statistics:")
print(
    data["word_count"].describe()
)


# -------------------- 24. 20K REQUIREMENT CHECK -----------------

print("\n" + "=" * 70)
print("20,000+ COMMENT REQUIREMENT")
print("=" * 70)

if len(data) >= 20000:

    print(
        f"✓ PASS: Dataset contains {len(data):,} comments."
    )

    print(
        "✓ Dataset satisfies the required 20,000+ comment target."
    )

else:

    print(
        f"⚠ WARNING: Dataset contains only {len(data):,} comments."
    )

    print(
        "⚠ Dataset does NOT currently satisfy the 20,000+ requirement."
    )


# -------------------- 25. SENTIMENT CHECK -----------------------

if "sentiment" in data.columns:

    print("\n" + "=" * 70)
    print("FINAL SENTIMENT LABELS")
    print("=" * 70)

    print(
        data["sentiment"]
        .value_counts()
        .to_string()
    )


# -------------------- 26. LANGUAGE CHECK -------------------------

if language_column:

    print("\n" + "=" * 70)
    print("FINAL LANGUAGE VALUES")
    print("=" * 70)

    print(
        data[language_column]
        .fillna("Unknown")
        .astype(str)
        .value_counts()
        .to_string()
    )


# -------------------- 27. MOST FREQUENT WORDS --------------------

all_words = []

for tokens in data["tokens_clean"]:
    all_words.extend(tokens)

word_frequency = Counter(all_words)

most_common_words = word_frequency.most_common(30)

word_frequency_df = pd.DataFrame(
    most_common_words,
    columns=["Word", "Frequency"]
)

print("\n" + "=" * 70)
print("TOP 30 MOST FREQUENT WORDS")
print("=" * 70)

display(word_frequency_df)


# -------------------- 28. SENTIMENT VISUALIZATION ---------------

if "sentiment" in data.columns:

    sentiment_counts = (
        data["sentiment"]
        .value_counts()
    )

    plt.figure(figsize=(8, 5))

    sentiment_counts.plot(
        kind="bar"
    )

    plt.title(
        "Sentiment Distribution of YouTube Comments"
    )

    plt.xlabel("Sentiment")
    plt.ylabel("Number of Comments")

    plt.xticks(rotation=0)

    plt.tight_layout()
    plt.show()


# -------------------- 29. COMMENT LENGTH VISUALIZATION ----------

plt.figure(figsize=(8, 5))

plt.hist(
    data["word_count"],
    bins=30
)

plt.title(
    "Distribution of YouTube Comment Length"
)

plt.xlabel("Number of Words")
plt.ylabel("Number of Comments")

plt.tight_layout()
plt.show()


# -------------------- 30. TRAIN / TEST SPLIT ---------------------

if "sentiment" in data.columns:

    # Remove missing sentiment labels
    data = data[
        data["sentiment"].str.strip() != ""
    ].copy()

    X = data["processed_text"]
    y = data["sentiment"]

    # Check whether stratification is possible
    sentiment_counts = y.value_counts()

    if len(sentiment_counts) >= 2 and sentiment_counts.min() >= 2:

        X_train, X_test, y_train, y_test = train_test_split(
            X,
            y,
            test_size=0.20,
            random_state=42,
            stratify=y
        )

    else:

        X_train, X_test, y_train, y_test = train_test_split(
            X,
            y,
            test_size=0.20,
            random_state=42
        )

    print("\n" + "=" * 70)
    print("TRAIN / TEST SPLIT")
    print("=" * 70)

    print("Training samples:", len(X_train))
    print("Testing samples :", len(X_test))


# -------------------- 31. TF-IDF FEATURE EXTRACTION ------------- 

if "sentiment" in data.columns:

    print("\n" + "=" * 70)
    print("TF-IDF FEATURE EXTRACTION")
    print("=" * 70)

    tfidf = TfidfVectorizer(
        max_features=10000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95,
        sublinear_tf=True
    )

    # Fit ONLY on training data
    X_train_tfidf = tfidf.fit_transform(
        X_train
    )

    # Transform test data
    X_test_tfidf = tfidf.transform(
        X_test
    )

    print(
        "Training TF-IDF shape:",
        X_train_tfidf.shape
    )

    print(
        "Testing TF-IDF shape :",
        X_test_tfidf.shape
    )

    print(
        "Number of TF-IDF features:",
        len(tfidf.get_feature_names_out())
    )


# -------------------- 32. SAVE PROCESSED DATASET -----------------

output_folder = "."

processed_file = (
    "final_english_hinglish_youtube_dataset.csv"
)

data.to_csv(
    processed_file,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 70)
print("FILES SAVED")
print("=" * 70)

print(
    "✓ Processed dataset:",
    processed_file
)


# -------------------- 33. SAVE TRAIN/TEST DATA ------------------

if "sentiment" in data.columns:

    train_data = data.loc[
        X_train.index
    ].copy()

    test_data = data.loc[
        X_test.index
    ].copy()

    train_data.to_csv(
        "train_dataset.csv",
        index=False,
        encoding="utf-8-sig"
    )

    test_data.to_csv(
        "test_dataset.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print("✓ Training dataset: train_dataset.csv")
    print("✓ Testing dataset : test_dataset.csv")


print("NLP PREPROCESSING COMPLETED SUCCESSFULLY")

print(f"""
Original dataset rows : {len(df):,}
Final dataset rows    : {len(data):,}

Text column           : {text_column}
Sentiment column      : {sentiment_column if sentiment_column else 'Not detected'}
Language column       : {language_column if language_column else 'Not detected'}

Output:
final_english_hinglish_youtube_dataset.csv
train_dataset.csv
test_dataset.csv
""")

print("=" * 70)
print("NEXT STEP: MACHINE LEARNING MODEL TRAINING")
print("=" * 70)
```
